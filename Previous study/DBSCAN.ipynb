```python
import pandas as pd
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('data/previous_data_2.csv')

data_clean = data.dropna()

data_clean = data_clean.select_dtypes(include=[float, int])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(data_clean)

data_clean.head()
```

```python
dbscan = DBSCAN(eps=0.5, min_samples=5)  # Puedes ajustar eps y min_samples
clusters = dbscan.fit_predict(X_scaled)

data_clean['Cluster'] = clusters

print(f"Puntos marcados como ruido: {(clusters == -1).sum()}")

plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=clusters, cmap='plasma')
plt.title('Agrupamiento DBSCAN')
plt.xlabel('Característica 1 (escalada)')
plt.ylabel('Característica 2 (escalada)')
plt.show()
```
