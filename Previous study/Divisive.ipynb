```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
```

```python
data = pd.read_csv('data/previous_data_2.csv')

data_clean = data.dropna()

data_clean = data_clean.select_dtypes(include=[float, int])
```

```python
Z = linkage(data_clean, method='ward')  

plt.figure(figsize=(10, 7))
plt.title('Dendrograma de Clustering Jerárquico Divisivo')
dendrogram(Z)
plt.show()

clusters = fcluster(Z, t=5, criterion='distance')

data_clean['Cluster'] = clusters

print(data_clean[['Cluster']].value_counts())

plt.scatter(data_clean.iloc[:, 0], data_clean.iloc[:, 1], c=clusters, cmap='prism')
plt.title('Clustering Jerárquico (Divisivo)')
plt.xlabel('Característica 1')
plt.ylabel('Característica 2')
plt.show()
```
